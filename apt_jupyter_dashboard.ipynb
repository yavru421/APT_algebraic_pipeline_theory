# APT Jupyter Dashboard - Pure Algebraic Pipeline Theory Implementation

## 🚀 APT METHODOLOGY IN JUPYTER

**Pipeline Equation:** `Y_dashboard = m6(m5(m4(m3(m2(m1(X_startup)))))`

**Contract:**
- **X_startup**: `{context_dirs, server_config, ui_config}`
- **Y_dashboard**: `{functional_ui, connected_server, persistent_state}`
- **All operations**: Follow strict APT m1→m2→m3→m4→m5→m6 patterns
- **Algebraic notation**: Every transformation explicitly defined
- **Traceability**: Full execution trace with timestamps and error handling

**Modules:**
- **m1**: System scan and validation
- **m2**: UI component construction  
- **m3**: Server connection management
- **m4**: Pipeline execution engine
- **m5**: State persistence
- **m6**: Status and UI updates

---

*This notebook recreates the APT Flet Dashboard and MCP Server Enterprise functionality using pure APT methodology in Jupyter Labs.*

## 1. APT Module m1: System Initialization

**Algebraic Contract:**
- **Input**: `x_startup = {context_dirs, server_url, ui_config}`
- **Output**: `y1_system_state = {platform_info, context_files, init_success}`
- **Errors**: APTError with recovery actions
- **Success**: Complete system state ready for UI construction

**Equation**: `y1 = m1(x_startup)`

````python
# ============================================================================
# APT MODULE DEFINITIONS - JUPYTER IMPLEMENTATION
# ============================================================================

import asyncio
import datetime
import json
import logging
import os
import platform
import time
from pathlib import Path
from typing import Dict, Any, List, Optional, Tuple

# Jupyter-specific imports
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output
import nest_asyncio

# HTTP and async
import httpx

# Enable nested event loops for Jupyter
nest_asyncio.apply()

# Configure logging for Jupyter
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
apt_logger = logging.getLogger('apt_jupyter')

class APTModule:
    """Base class for all APT modules with algebraic contracts"""

    def __init__(self, module_name: str):
        self.module_name = module_name
        self.execution_log = []

    def log_execution(self, inputs: Dict[str, Any], outputs: Dict[str, Any], success: bool):
        """Log module execution for traceability"""
        self.execution_log.append({
            "timestamp": datetime.datetime.now().isoformat(),
            "module": self.module_name,
            "inputs": inputs,
            "outputs": outputs,
            "success": success
        })

class APTError:
    """Algebraic error representation instead of exceptions"""

    def __init__(self, module: str, error_type: str, message: str, recovery_action: str = None):
        self.module = module
        self.error_type = error_type
        self.message = message
        self.recovery_action = recovery_action
        self.timestamp = datetime.datetime.now().isoformat()

    def to_dict(self) -> Dict[str, Any]:
        return {
            "module": self.module,
            "error_type": self.error_type,
            "message": self.message,
            "recovery_action": self.recovery_action,
            "timestamp": self.timestamp
        }

print("✅ APT Module system initialized")
````

````python
# ============================================================================
# APT MODULE m1: SYSTEM INITIALIZATION
# ============================================================================

class M1_SystemInit(APTModule):
    """
    APT Module m1: System Initialization

    Contract:
      Inputs: x_startup = {context_dirs, server_url, ui_config}
      Outputs: y1_system_state = {platform_info, context_files, init_success}
      Errors: APTError with recovery actions
      Success: Complete system state ready for UI construction

    Algebraic: y1 = m1(x_startup)
    """

    def __init__(self):
        super().__init__("m1_system_init")

    async def execute(self, x_startup: Dict[str, Any]) -> Tuple[Dict[str, Any], Optional[APTError]]:
        """Execute m1 with algebraic error handling"""
        inputs = {"x_startup": x_startup}

        try:
            apt_logger.info("m1_system_scan: Starting system validation")

            # Extract startup parameters
            context_dirs = x_startup.get("context_dirs", ["context_files", "context", "docs"])
            server_url = x_startup.get("server_url", "http://localhost:8000")
            ui_config = x_startup.get("ui_config", {})

            # Scan context directories
            context_files = {}
            for dirname in context_dirs:
                dir_path = Path(os.getcwd()) / dirname
                if dir_path.exists() and dir_path.is_dir():
                    files = []
                    for file_path in dir_path.rglob("*"):
                        if file_path.is_file() and file_path.suffix in [".txt", ".md", ".json", ".py"]:
                            files.append(str(file_path.relative_to(os.getcwd())))
                    if files:
                        context_files[dirname] = files

            # Build system state
            y1_system_state = {
                "platform_info": {
                    "os": os.name,
                    "platform": platform.platform(),
                    "python_version": platform.python_version(),
                    "cwd": os.getcwd()
                },
                "context_files": context_files,
                "server_url": server_url,
                "ui_config": ui_config,
                "init_timestamp": time.time(),
                "init_success": True
            }

            self.log_execution(inputs, {"y1_system_state": y1_system_state}, True)
            apt_logger.info(f"m1_system_scan: Complete - found {len(context_files)} directories")
            return y1_system_state, None

        except Exception as e:
            apt_logger.error(f"m1_system_scan: Failed - {str(e)}")
            error = APTError(
                "m1_system_init",
                "initialization_error",
                f"System scan failed: {str(e)}",
                "Check file permissions and directory structure"
            )
            return {}, error

# Global m1 instance
m1_system_init = M1_SystemInit()

print("✅ APT Module m1 (System Initialization) loaded")
````

## 2. APT Module m2: UI Component Construction

**Algebraic Contract:**
- **Input**: `y1_system_state`
- **Output**: `y2_ui_components = {status_bar, input_field, output_field, control_buttons}`
- **Errors**: APTError for UI construction failures
- **Success**: Complete UI component tree ready for display

**Equation**: `y2 = m2(y1)`

````python
# ============================================================================
# APT MODULE m2: UI COMPONENT CONSTRUCTION
# ============================================================================

class M2_UIConstruction(APTModule):
    """
    APT Module m2: UI Component Construction

    Contract:
      Inputs: y1_system_state
      Outputs: y2_ui_components = {status_bar, input_field, output_field, control_buttons}
      Errors: APTError for UI construction failures
      Success: Complete UI component tree ready for display

    Algebraic: y2 = m2(y1)
    """

    def __init__(self):
        super().__init__("m2_ui_construction")
        self.ui_components = {}

    def execute(self, y1_system_state: Dict[str, Any]) -> Tuple[Dict[str, Any], Optional[APTError]]:
        """Execute m2 with UI component construction"""
        inputs = {"y1_system_state": y1_system_state}

        try:
            apt_logger.info("m2_ui_construction: Building Jupyter UI components")

            # Status display widget
            status_output = widgets.Output()
            with status_output:
                print("🔄 Initializing APT Dashboard...")

            # Instruction input
            instruction_input = widgets.Text(
                value='',
                placeholder='Enter APT instruction: e.g., "analyze context_files"',
                description='Instruction:',
                layout=widgets.Layout(width='70%')
            )

            # Execute button
            execute_button = widgets.Button(
                description='Execute Pipeline',
                button_style='success',
                tooltip='Execute the APT pipeline'
            )

            # Pipeline equation display
            pipeline_display = widgets.Textarea(
                value="Y = m6(m5(m4(m3(m2(m1(X))))))",
                description='Pipeline:',
                layout=widgets.Layout(width='100%', height='40px'),
                disabled=True
            )

            # Results output
            results_output = widgets.Output(
                layout=widgets.Layout(width='100%', height='300px', border='1px solid #ccc')
            )

            # Context files display
            context_info = y1_system_state.get("context_files", {})
            context_display = widgets.Textarea(
                value=f"Context directories found: {len(context_info)}\n" +
                      "\n".join([f"{k}: {len(v)} files" for k, v in context_info.items()]),
                description='Context:',
                layout=widgets.Layout(width='100%', height='100px'),
                disabled=True
            )

            # Control buttons
            refresh_button = widgets.Button(
                description='Refresh Status',
                button_style='info',
                tooltip='Refresh system and server status'
            )

            clear_button = widgets.Button(
                description='Clear Results',
                button_style='warning',
                tooltip='Clear all results'
            )

            export_button = widgets.Button(
                description='Export State',
                button_style='',
                tooltip='Export current dashboard state'
            )

            # Build UI component tree
            y2_ui_components = {
                "status_output": status_output,
                "instruction_input": instruction_input,
                "execute_button": execute_button,
                "pipeline_display": pipeline_display,
                "results_output": results_output,
                "context_display": context_display,
                "refresh_button": refresh_button,
                "clear_button": clear_button,
                "export_button": export_button,
                "context_info": f"Context files: {len(context_info)} directories",
                "construction_success": True
            }

            self.ui_components = y2_ui_components
            self.log_execution(inputs, {"y2_ui_components": "jupyter_widgets_constructed"}, True)
            apt_logger.info("m2_ui_construction: Jupyter UI components built successfully")
            return y2_ui_components, None

        except Exception as e:
            apt_logger.error(f"m2_ui_construction: Failed - {str(e)}")
            error = APTError(
                "m2_ui_construction",
                "ui_construction_error",
                f"UI construction failed: {str(e)}",
                "Check Jupyter widgets installation and compatibility"
            )
            return {}, error

# Global m2 instance
m2_ui_construction = M2_UIConstruction()

print("✅ APT Module m2 (UI Construction) loaded")
````

## 3. APT Module m3: Server Connection

**Algebraic Contract:**
- **Input**: `y2_ui_components`
- **Output**: `y3_connected_system = {server_status, connection_time, api_endpoints}`
- **Errors**: APTError for connection failures with recovery actions
- **Success**: Established server connection ready for pipeline execution

**Equation**: `y3 = m3(y2)`

````python
# ============================================================================
# APT MODULE m3: SERVER CONNECTION
# ============================================================================

class M3_ServerConnection(APTModule):
    """
    APT Module m3: Server Connection Management

    Contract:
      Inputs: y2_ui_components
      Outputs: y3_connected_system = {server_status, connection_time, api_endpoints}
      Errors: APTError for connection failures with recovery actions
      Success: Established server connection ready for pipeline execution

    Algebraic: y3 = m3(y2)
    """

    def __init__(self):
        super().__init__("m3_server_connection")
        self.is_connected = False
        self.last_check = 0

    async def execute(self, y2_ui_components: Dict[str, Any]) -> Tuple[Dict[str, Any], Optional[APTError]]:
        """Execute m3 with algebraic server connection"""
        inputs = {"y2_ui_components": "ui_components_ready"}

        try:
            apt_logger.info("m3_server_connection: Attempting server connection")

            # Extract server URL from system state or use default
            server_url = "http://localhost:8000"

            # Algebraic connection attempt
            connection_start = time.time()
            server_status = await self._attempt_connection(server_url)
            connection_time = time.time() - connection_start

            if server_status["connected"]:
                y3_connected_system = {
                    "server_url": server_url,
                    "connection_status": "connected",
                    "connection_time": connection_time,
                    "api_endpoints": {
                        "health": f"{server_url}/health",
                        "chat": f"{server_url}/v1/chat/completions",
                        "pipeline": f"{server_url}/v1/pipeline/execute",
                        "vector_search": f"{server_url}/v1/vector/search"
                    },
                    "server_info": server_status.get("server_info", {}),
                    "connection_success": True
                }
                self.is_connected = True
                apt_logger.info(f"m3_server_connection: Connected successfully in {connection_time:.2f}s")
            else:
                y3_connected_system = {
                    "server_url": server_url,
                    "connection_status": "disconnected",
                    "connection_time": connection_time,
                    "error": server_status.get("error", "Unknown connection error"),
                    "connection_success": False
                }
                self.is_connected = False
                apt_logger.warning(f"m3_server_connection: Connection failed - {server_status.get('error', 'Unknown error')}")

            self.log_execution(inputs, {"y3_connected_system": y3_connected_system}, y3_connected_system["connection_success"])
            return y3_connected_system, None

        except Exception as e:
            apt_logger.error(f"m3_server_connection: Failed - {str(e)}")
            error = APTError(
                "m3_server_connection",
                "connection_error",
                f"Server connection failed: {str(e)}",
                "Ensure MCP server is running on localhost:8000"
            )
            return {}, error

    async def _attempt_connection(self, server_url: str) -> Dict[str, Any]:
        """Attempt server connection with proper timeout"""
        try:
            async with httpx.AsyncClient(timeout=5.0) as client:
                response = await client.get(f"{server_url}/health")

                if response.status_code == 200:
                    server_info = response.json()
                    return {
                        "connected": True,
                        "server_info": server_info,
                        "status_code": response.status_code,
                        "response_time": response.elapsed.total_seconds()
                    }
                else:
                    return {
                        "connected": False,
                        "error": f"HTTP {response.status_code}",
                        "status_code": response.status_code
                    }

        except httpx.TimeoutException:
            return {
                "connected": False,
                "error": "Connection timeout",
                "timeout": True
            }
        except httpx.ConnectError as e:
            return {
                "connected": False,
                "error": f"Connection refused: {str(e)}",
                "connection_refused": True
            }
        except Exception as e:
            return {
                "connected": False,
                "error": f"Unexpected error: {str(e)}"
            }

# Global m3 instance
m3_server_connection = M3_ServerConnection()

print("✅ APT Module m3 (Server Connection) loaded")
````

## 4. APT Module m4: Pipeline Execution Engine

**Algebraic Contract:**
- **Input**: `y3_connected_system + x_instruction`
- **Output**: `y4_results = {instruction, execution_trace, server_response, success}`
- **Errors**: APTError for execution failures
- **Success**: Complete pipeline execution with full trace

**Equation**: `y4 = m4(y3, x_instruction)`

````python
# ============================================================================
# APT MODULE m4: PIPELINE EXECUTION ENGINE
# ============================================================================

class M4_PipelineExecution(APTModule):
    """
    APT Module m4: Pipeline Execution Engine

    Contract:
      Inputs: y3_connected_system + x_instruction
      Outputs: y4_results = {instruction, execution_trace, server_response, success}
      Errors: APTError for execution failures
      Success: Complete pipeline execution with full trace

    Algebraic: y4 = m4(y3, x_instruction)
    """

    def __init__(self):
        super().__init__("m4_pipeline_execution")
        self.execution_history = []

    async def execute(self, y3_connected_system: Dict[str, Any], x_instruction: str) -> Tuple[Dict[str, Any], Optional[APTError]]:
        """Execute m4 with full pipeline execution"""
        inputs = {"y3_connected_system": "system_ready", "x_instruction": x_instruction}

        try:
            apt_logger.info(f"m4_pipeline_execution: Executing instruction - {x_instruction[:50]}...")

            if not y3_connected_system.get("connection_success", False):
                error = APTError(
                    "m4_pipeline_execution",
                    "connection_error",
                    "Server not connected",
                    "Execute m3 first to establish server connection"
                )
                return {}, error

            # Execute pipeline on server
            server_url = y3_connected_system["server_url"]
            execution_trace = []

            # Step 1: Parse instruction
            execution_trace.append({
                "step": "parse_instruction",
                "timestamp": datetime.datetime.now().isoformat(),
                "action": "Parsing APT instruction"
            })

            # Step 2: Execute on server
            execution_trace.append({
                "step": "server_execution",
                "timestamp": datetime.datetime.now().isoformat(),
                "action": f"Executing on {server_url}"
            })

            try:
                async with httpx.AsyncClient(timeout=30.0) as client:
                    # Try pipeline execution first
                    pipeline_response = await client.post(
                        f"{server_url}/v1/pipeline/execute",
                        json={"instruction": x_instruction, "cache_results": True}
                    )

                    if pipeline_response.status_code == 200:
                        server_response = pipeline_response.json()
                        execution_trace.append({
                            "step": "pipeline_complete",
                            "timestamp": datetime.datetime.now().isoformat(),
                            "action": "Pipeline execution successful",
                            "response": server_response
                        })

                        y4_results = {
                            "instruction": x_instruction,
                            "execution_trace": execution_trace,
                            "server_response": server_response,
                            "result": server_response.get("result", "No result"),
                            "success": True,
                            "execution_time": time.time()
                        }
                    else:
                        # Fallback to chat completion
                        execution_trace.append({
                            "step": "fallback_chat",
                            "timestamp": datetime.datetime.now().isoformat(),
                            "action": "Pipeline failed, trying chat completion"
                        })

                        chat_response = await client.post(
                            f"{server_url}/v1/chat/completions",
                            json={
                                "prompt": x_instruction,
                                "model": "llama-3.3-70b",
                                "max_tokens": 2048
                            }
                        )

                        if chat_response.status_code == 200:
                            server_response = chat_response.json()
                            execution_trace.append({
                                "step": "chat_complete",
                                "timestamp": datetime.datetime.now().isoformat(),
                                "action": "Chat completion successful",
                                "response": server_response
                            })

                            y4_results = {
                                "instruction": x_instruction,
                                "execution_trace": execution_trace,
                                "server_response": server_response,
                                "result": server_response.get("response", "No response"),
                                "success": True,
                                "execution_time": time.time(),
                                "fallback_used": True
                            }
                        else:
                            raise Exception(f"Server error: {chat_response.status_code}")

            except Exception as e:
                execution_trace.append({
                    "step": "execution_failed",
                    "timestamp": datetime.datetime.now().isoformat(),
                    "action": f"Execution failed: {str(e)}",
                    "error": str(e)
                })

                error = APTError(
                    "m4_pipeline_execution",
                    "execution_error",
                    f"Pipeline execution failed: {str(e)}",
                    "Check server connection and instruction format"
                )
                return {}, error

            # Add to execution history
            self.execution_history.append({
                "id": str(int(time.time())),
                "timestamp": datetime.datetime.now().isoformat(),
                "instruction": x_instruction,
                "result": y4_results,
                "success": True
            })

            self.log_execution(inputs, {"y4_results": "execution_complete"}, True)
            apt_logger.info("m4_pipeline_execution: Pipeline execution completed successfully")
            return y4_results, None

        except Exception as e:
            apt_logger.error(f"m4_pipeline_execution: Failed - {str(e)}")
            error = APTError(
                "m4_pipeline_execution",
                "execution_error",
                f"Pipeline execution failed: {str(e)}",
                "Check server connection and try again"
            )
            return {}, error

# Global m4 instance
m4_pipeline_execution = M4_PipelineExecution()

print("✅ APT Module m4 (Pipeline Execution) loaded")
````

## 5. APT Module m5: State Persistence

**Algebraic Contract:**
- **Input**: `y4_results`
- **Output**: `y5_persisted_state = {results, saved_state, persistence_success}`
- **Errors**: APTError for persistence failures
- **Success**: State successfully persisted with backup

**Equation**: `y5 = m5(y4)`

````python
# ============================================================================
# APT MODULE m5: STATE PERSISTENCE
# ============================================================================

class M5_StatePersistence(APTModule):
    """
    APT Module m5: State Persistence Management

    Contract:
      Inputs: y4_results
      Outputs: y5_persisted_state = {results, saved_state, persistence_success}
      Errors: APTError for persistence failures
      Success: State successfully persisted with backup

    Algebraic: y5 = m5(y4)
    """

    def __init__(self):
        super().__init__("m5_state_persistence")
        self.state_file = Path(os.getcwd()) / ".apt_jupyter_state.json"
        self.current_state = {}

    async def execute(self, y4_results: Dict[str, Any]) -> Tuple[Dict[str, Any], Optional[APTError]]:
        """Execute m5 with state persistence"""
        inputs = {"y4_results": "execution_completed"}

        try:
            apt_logger.info("m5_state_persistence: Saving execution state")

            # Load existing state
            existing_state = await self._load_existing_state()

            # Update state with new results
            execution_record = {
                "timestamp": datetime.datetime.now().isoformat(),
                "instruction": y4_results.get("instruction", ""),
                "result": y4_results.get("result", ""),
                "success": y4_results.get("success", False),
                "execution_trace": y4_results.get("execution_trace", [])
            }

            # Add to execution history
            if "execution_history" not in existing_state:
                existing_state["execution_history"] = []

            existing_state["execution_history"].append(execution_record)

            # Keep only last 50 executions
            existing_state["execution_history"] = existing_state["execution_history"][-50:]

            # Update metadata
            existing_state["last_updated"] = datetime.datetime.now().isoformat()
            existing_state["total_executions"] = len(existing_state["execution_history"])

            # Save state
            success = await self._save_state(existing_state)

            if success:
                y5_persisted_state = {
                    "results": y4_results,
                    "saved_state": existing_state,
                    "persistence_success": True,
                    "state_file": str(self.state_file)
                }

                self.current_state = existing_state
                self.log_execution(inputs, {"y5_persisted_state": "state_saved"}, True)
                apt_logger.info("m5_state_persistence: State saved successfully")
                return y5_persisted_state, None
            else:
                error = APTError(
                    "m5_state_persistence",
                    "save_error",
                    "Failed to save state to file",
                    "Check file permissions and disk space"
                )
                return {}, error

        except Exception as e:
            apt_logger.error(f"m5_state_persistence: Failed - {str(e)}")
            error = APTError(
                "m5_state_persistence",
                "persistence_error",
                f"State persistence failed: {str(e)}",
                "Check file system and try again"
            )
            return {}, error

    async def _load_existing_state(self) -> Dict[str, Any]:
        """Load existing state file"""
        try:
            if self.state_file.exists():
                with open(self.state_file, 'r', encoding='utf-8') as f:
                    return json.load(f)
            else:
                return {
                    "created": datetime.datetime.now().isoformat(),
                    "execution_history": [],
                    "total_executions": 0
                }
        except Exception as e:
            apt_logger.warning(f"Failed to load existing state: {str(e)}")
            return {}

    async def _save_state(self, state_data: Dict[str, Any]) -> bool:
        """Save state to persistent storage"""
        try:
            # Create backup if file exists
            if self.state_file.exists():
                backup_file = self.state_file.with_suffix('.backup.json')
                self.state_file.rename(backup_file)

            # Save new state
            with open(self.state_file, 'w', encoding='utf-8') as f:
                json.dump(state_data, f, indent=2, ensure_ascii=False)

            apt_logger.info(f"State saved to {self.state_file}")
            return True

        except Exception as e:
            apt_logger.error(f"Failed to save state: {str(e)}")
            return False

    def get_execution_history(self) -> List[Dict[str, Any]]:
        """Get execution history"""
        return self.current_state.get("execution_history", [])

# Global m5 instance
m5_state_persistence = M5_StatePersistence()

print("✅ APT Module m5 (State Persistence) loaded")
````

## 6. APT Module m6: Status and UI Update

**Algebraic Contract:**
- **Input**: `y5_persisted_state`
- **Output**: `Y_dashboard = {final_ui_state, status_updated, dashboard_ready}`
- **Errors**: APTError for UI update failures
- **Success**: Complete dashboard with updated status and results

**Equation**: `Y = m6(y5)`

````python
# ============================================================================
# APT MODULE m6: STATUS AND UI UPDATE
# ============================================================================

class M6_StatusUpdate(APTModule):
    """
    APT Module m6: Status and UI Update

    Contract:
      Inputs: y5_persisted_state
      Outputs: Y_dashboard = {final_ui_state, status_updated, dashboard_ready}
      Errors: APTError for UI update failures
      Success: Complete dashboard with updated status and results

    Algebraic: Y = m6(y5)
    """

    def __init__(self):
        super().__init__("m6_status_update")

    def execute(self, y5_persisted_state: Dict[str, Any], ui_components: Dict[str, Any]) -> Tuple[Dict[str, Any], Optional[APTError]]:
        """Execute m6 with UI status updates"""
        inputs = {"y5_persisted_state": "state_persisted"}

        try:
            apt_logger.info("m6_status_update: Updating dashboard UI")

            # Extract results from persisted state
            results = y5_persisted_state.get("results", {})
            saved_state = y5_persisted_state.get("saved_state", {})

            # Update status display
            status_message = self._generate_status_message(results, saved_state)
            self._update_status_display(ui_components, status_message)

            # Update results display
            result_content = self._format_results(results)
            self._update_results_display(ui_components, result_content)

            # Update pipeline equation if available
            if results.get("equation"):
                self._update_pipeline_display(ui_components, results["equation"])

            # Update execution history
            history_summary = self._generate_history_summary(saved_state)
            self._update_history_display(ui_components, history_summary)

            Y_dashboard = {
                "final_ui_state": "updated",
                "status_updated": True,
                "dashboard_ready": True,
                "last_update": datetime.datetime.now().isoformat(),
                "results_displayed": bool(results),
                "history_available": len(saved_state.get("execution_history", [])) > 0
            }

            self.log_execution(inputs, {"Y_dashboard": Y_dashboard}, True)
            apt_logger.info("m6_status_update: Dashboard UI updated successfully")
            return Y_dashboard, None

        except Exception as e:
            apt_logger.error(f"m6_status_update: Failed - {str(e)}")
            error = APTError(
                "m6_status_update",
                "ui_update_error",
                f"UI update failed: {str(e)}",
                "Check UI components and try again"
            )
            return {}, error

    def _generate_status_message(self, results: Dict[str, Any], saved_state: Dict[str, Any]) -> str:
        """Generate status message based on current state"""
        if results.get("success"):
            total_executions = saved_state.get("total_executions", 0)
            return f"✅ Pipeline executed successfully | Total executions: {total_executions}"
        elif results.get("error"):
            return f"❌ Execution failed: {results['error']}"
        else:
            return "🔄 Ready for execution"

    def _update_status_display(self, ui_components: Dict[str, Any], message: str):
        """Update status display widget"""
        status_output = ui_components.get("status_output")
        if status_output:
            with status_output:
                clear_output(wait=True)
                print(message)

    def _format_results(self, results: Dict[str, Any]) -> str:
        """Format results for display"""
        if not results:
            return "No results available"

        formatted = f"📝 Instruction: {results.get('instruction', 'N/A')}\n\n"

        if results.get("success"):
            formatted += f"✅ Result: {results.get('result', 'No result')}\n\n"
        else:
            formatted += f"❌ Error: {results.get('error', 'Unknown error')}\n\n"

        # Add execution trace
        trace = results.get("execution_trace", [])
        if trace:
            formatted += "🔍 Execution Trace:\n"
            for step in trace:
                formatted += f"  • {step.get('step', 'unknown')}: {step.get('action', '')}\n"
            formatted += "\n"

        formatted += f"🕒 Executed: {datetime.datetime.now().strftime('%H:%M:%S')}"
        return formatted

    def _update_results_display(self, ui_components: Dict[str, Any], content: str):
        """Update results display widget"""
        results_output = ui_components.get("results_output")
        if results_output:
            with results_output:
                clear_output(wait=True)
                print(content)

    def _update_pipeline_display(self, ui_components: Dict[str, Any], equation: str):
        """Update pipeline equation display"""
        pipeline_display = ui_components.get("pipeline_display")
        if pipeline_display:
            pipeline_display.value = equation

    def _generate_history_summary(self, saved_state: Dict[str, Any]) -> str:
        """Generate execution history summary"""
        history = saved_state.get("execution_history", [])
        if not history:
            return "No execution history available"

        recent = history[-5:]  # Last 5 executions
        summary = f"📊 Recent Executions ({len(history)} total):\n\n"

        for i, execution in enumerate(reversed(recent), 1):
            timestamp = execution.get("timestamp", "")[:19]  # YYYY-MM-DDTHH:MM:SS
            instruction = execution.get("instruction", "")[:50]
            success = "✅" if execution.get("success") else "❌"
            summary += f"{i}. {success} {timestamp} | {instruction}...\n"

        return summary

    def _update_history_display(self, ui_components: Dict[str, Any], summary: str):
        """Update history display (could be added to UI components)"""
        # For now, just log the summary
        apt_logger.info(f"Execution history summary: {summary[:100]}...")

# Global m6 instance
m6_status_update = M6_StatusUpdate()

print("✅ APT Module m6 (Status Update) loaded")
````

## 7. APT Pipeline Orchestrator

**Algebraic Contract:**
- **Input**: `X_startup = {context_dirs, server_config, ui_config}`
- **Output**: `Y_dashboard = m6(m5(m4(m3(m2(m1(X_startup)))))`
- **Modules**: Orchestrates execution of all APT modules in sequence
- **Success**: Complete dashboard pipeline execution

**Equation**: `Y = m6(m5(m4(m3(m2(m1(X))))))`

````python
# ============================================================================
# APT PIPELINE ORCHESTRATOR
# ============================================================================

class APTPipelineOrchestrator:
    """
    APT Pipeline Orchestrator - Executes the complete algebraic pipeline

    Pipeline Equation: Y = m6(m5(m4(m3(m2(m1(X)))))
    """

    def __init__(self):
        self.m1 = m1_system_init
        self.m2 = m2_ui_construction
        self.m3 = m3_server_connection
        self.m4 = m4_pipeline_execution
        self.m5 = m5_state_persistence
        self.m6 = m6_status_update

        self.execution_errors = []
        self.current_state = None
        self.ui_components = None

    async def execute_full_pipeline(self) -> Dict[str, Any]:
        """Execute the complete APT pipeline: Y = m6(m5(m4(m3(m2(m1(X)))))"""
        apt_logger.info("🚀 Starting APT Pipeline Execution: Y = m6(m5(m4(m3(m2(m1(X))))))")

        # X_startup: Initial input
        X_startup = {
            "context_dirs": ["context_files", "context", "docs"],
            "server_url": "http://localhost:8000",
            "ui_config": {"theme": "jupyter", "debug": False}
        }

        try:
            # m1: System initialization
            apt_logger.info("📡 m1: System initialization")
            y1, error1 = await self.m1.execute(X_startup)
            if error1:
                self.execution_errors.append(error1)
                apt_logger.error(f"m1 failed: {error1.message}")
                return {"pipeline_state": {}, "errors": [e.to_dict() for e in self.execution_errors], "success": False}

            # m2: UI construction
            apt_logger.info("🎨 m2: UI component construction")
            y2, error2 = self.m2.execute(y1)
            if error2:
                self.execution_errors.append(error2)
                apt_logger.error(f"m2 failed: {error2.message}")
                return {"pipeline_state": y1, "errors": [e.to_dict() for e in self.execution_errors], "success": False}

            self.ui_components = y2

            # m3: Server connection
            apt_logger.info("🔗 m3: Server connection")
            y3, error3 = await self.m3.execute(y2)
            if error3:
                self.execution_errors.append(error3)
                apt_logger.warning(f"m3 failed: {error3.message}")
                # Continue with disconnected state

            # Store current state for instruction execution
            self.current_state = y3

            pipeline_result = {
                "pipeline_state": y3,
                "errors": [e.to_dict() for e in self.execution_errors],
                "success": len(self.execution_errors) == 0,
                "modules_executed": ["m1", "m2", "m3"],
                "ui_ready": True
            }

            apt_logger.info("✅ APT Pipeline initialization complete")
            return pipeline_result

        except Exception as e:
            apt_logger.error(f"Pipeline execution failed: {str(e)}")
            error = APTError("orchestrator", "pipeline_error", f"Pipeline failed: {str(e)}")
            self.execution_errors.append(error)
            return {"pipeline_state": {}, "errors": [e.to_dict() for e in self.execution_errors], "success": False}

    async def execute_instruction(self, instruction: str) -> Dict[str, Any]:
        """Execute instruction through m4→m5→m6"""
        if not self.current_state:
            return {"error": "Pipeline not initialized", "success": False}

        try:
            apt_logger.info(f"⚙️ Executing instruction: {instruction[:50]}...")

            # m4: Pipeline execution
            y4, error4 = await self.m4.execute(self.current_state, instruction)
            if error4:
                self.execution_errors.append(error4)
                return {"error": error4.message, "success": False}

            # m5: State persistence
            y5, error5 = await self.m5.execute(y4)
            if error5:
                self.execution_errors.append(error5)
                apt_logger.warning(f"m5 failed: {error5.message}")

            # m6: Status update
            Y_final, error6 = self.m6.execute(y5, self.ui_components or {})
            if error6:
                self.execution_errors.append(error6)
                apt_logger.warning(f"m6 failed: {error6.message}")

            result = {
                "result": y4.get("result", "No result"),
                "success": y4.get("success", False),
                "execution_trace": y4.get("execution_trace", []),
                "modules_executed": ["m4", "m5", "m6"]
            }

            apt_logger.info("✅ Instruction execution complete")
            return result

        except Exception as e:
            apt_logger.error(f"Instruction execution failed: {str(e)}")
            return {"error": str(e), "success": False}

    def get_ui_components(self) -> Dict[str, Any]:
        """Get UI components for display"""
        return self.ui_components or {}

    def get_execution_errors(self) -> List[Dict[str, Any]]:
        """Get execution errors"""
        return [e.to_dict() for e in self.execution_errors]

# Global orchestrator instance
apt_orchestrator = APTPipelineOrchestrator()

print("✅ APT Pipeline Orchestrator loaded")
````
